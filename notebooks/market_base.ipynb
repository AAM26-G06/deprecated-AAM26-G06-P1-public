## A. Adquisición del conjunto de datos y exploración inicial

**Objetivo:** Descargar el conjunto de datos directamente desde el repositorio del proyecto en la carpeta local `data/raw/` y verificar que la estructura de columnas y tipos de datos se haya cargado correctamente sin desfases en la cabecera.

```python
# 1. Rutas locales y URL oficial
PROJECT_ROOT = os.path.abspath(os.path.join(".."))
RAW_DATA_DIR = os.path.join(PROJECT_ROOT, "data", "raw")
LOCAL_CSV_PATH = os.path.join(RAW_DATA_DIR, "games.csv")
DATASET_URL = "https://github.com/narratech/market-base/releases/download/Release/games.csv"
import csv

# 2. Descarga automatizada a data/raw/
if not os.path.exists(LOCAL_CSV_PATH):
    os.makedirs(RAW_DATA_DIR, exist_ok=True)
    print("Descargando games.csv desde el repositorio...")
    urllib.request.urlretrieve(DATASET_URL, LOCAL_CSV_PATH)
    print(f"¡Descarga completada en {LOCAL_CSV_PATH}!")
else:
    print(f"El archivo ya existe en {LOCAL_CSV_PATH}.")


# Leer el CSV.
with open(LOCAL_CSV_PATH, "r", encoding="utf-8") as f:
    header = f.readline().strip() # Primera linea, encabezado. Strip para quitar espacios en blanco al principio y final.

goodHeader = header.replace("DiscountDLC count", "Discount,DLC count") # El archivo original no tenia coma entre ambas columnas. Lo cambiamos a mano.
headerList = goodHeader.split(",") # Encabezado corregido y separado por comas.

# Para cargar.
dfRaw = pd.read_csv(LOCAL_CSV_PATH,
                     skiprows = 1, # Como la primera linea es el encabezado y no son datos => la quitamos.
                     names = headerList, # Como no hemos cogido el encabezado entonces los ponemos a mano.
                     encoding = "utf-8") # Codificacion del archivo.

print("[M] Data frame creado.")
```

Se trata de un archivo .csv con datos sobre 125855 juegos de 40 categorías diferentes (AppID, Name, Release date, Estimated owners, etc.). ## B. Valoración de la calidad de datos, limpieza e ingeniería de características

**Objetivo:** Tratar valores nulos, transformar tipos de datos (como la conversión de cadenas de texto separadas por comas a listas iterables) y calcular nuevas métricas como la tasa de valoraciones positivas, aplicando un filtrado adecuado para evitar sesgos por bajo volumen de reseñas.

```python
# Tratamiento de nulos y conversión de tipos de datos (ej. cadenas a listas para géneros/tags)
# Escribe tu código aquí:

from datetime import datetime

# Creamos una copia del dataframe por si acaso y por buena practica. Modificamos la copia en vez del original.
df = dfRaw.copy()

# --- Limpiar el data frame: Quitamos columnas con muchos valores nulos o que no nos interesen.
columnasEliminar=["Movies", "Screenshots", "Notes", "Score rank", "Reviews", "Support url", "Support email", "Metacritic url", "Tags"]
# Quitar columnas que no interesen.
df = df.drop(columns=columnasEliminar) # o con "df.drop(columns=columnasEliminar, inplace=true)"

print(df["Release date"])
# --- Transformar tipos de datos
# -> Conversión de las fechas de tipo str a formato Date de Pandas.
df["Release date"] = pd.to_datetime(df["Release date"], # La que queremos convertir.
                                    format = "mixed",   # Para que pueda haber fechas en varios formatos. Pandas sabe interpretar la mayoria.
                                    errors = "coerce")  # Convierte las fechas invaliadas a NaT (Not a Time)

print(df["Release date"])

# Valores de texto que pueden estar nulos. Meter las que queramos
strColumns = ["Developers", "Publishers", "Genres", "Categories"]
unknowStrValue = "UnknowName"
for col in strColumns:
    df[col] = df[col].fillna(unknowStrValue) # Llena los valores ausentes con la etiqueta que queramos.

# -> Conversión de cadenas de texto separadas por comas a listas iterables.
def strToList(cadena):
    if isinstance(cadena, str) and cadena != unknowStrValue: # Comprobamos que la cadena sea str y que no tenga la etiqueta que le hemos puesto antes.
        return[item.strip() for item in cadena.split(",")] # Separamos las cadenas qeu tengan varios str separandolos habeindolos previamente separado por comas.
    else:
        return [] # Devolvemos una lista vacia si es la etiqueta o no es un str que significa que tampoco es valido.

strListsColumns = ["Genres", "Categories"] # Las columnas que queramos separar.
for col in strListsColumns:
    df[col] = df[col].apply(strToList) # Que aplique nuetra funcion

```

```python
# Crear la característica 'tasa_positividad' filtrando solo juegos con más de 50 reseñas totales

nReviews = df["Positive"] + df["Negative"]

minReviews = 50
notEnough = -1 # Valor a poner si no hay suficientes valoraciones.

# 3. Cálculo de la tasa individual para cada juego
df["tasa_positividad"] = np.where(nReviews > minReviews, # Solo si se cumple que tenga min 50 valoraciones.
                                (df["Positive"] / nReviews) * 100, # Tasa de positividad.
                                notEnough # Si no lo llenamos con nuestro valor.
                                ).round(2) # Para que solo tenga 2 decimales.

print(df[["Name", "Positive", "Negative", "tasa_positividad"]].head(50)) # Para comprobar.
```

> **Discusión**  
> *Explica el impacto del tratamiento de nulos y justifica cuantitativamente por qué es técnicamente necesario filtrar por un volumen mínimo de reseñas (50) para evitar distorsiones en la métrica de positividad.*

Si no se filtra por tamaño el número de reseñas con una tasa de positividad del 100% se dispara, y esto no es realista, conviene entonces tratar los datos para que no tengan ese sesgo, aplanando la curva de tasa de positividad. Se visualizan a continuación los resultados sin esta corrección.

```python
df["totalReviews"] = nReviews = df["Positive"] + df["Negative"]

minReviews = 0 # Sin filtro
notEnough = -1 # Valor a poner si no hay suficientes valoraciones.

df["tasa_positividad"] = np.where(nReviews > minReviews, # Solo si se cumple que tenga min 50 valoraciones.
                                (df["Positive"] / nReviews) * 100, # El valor que calculado.
                                notEnough # Sino lo llenamos con nuestro valor.
                                ).round(2) # Para que solo tenga 2 decimales.

print(df[["Name", "Positive", "Negative", "tasa_positividad"]].head(50)) # Para comprobar.

# Visualización de los datos
print(df[["tasa_positividad"]].value_counts())

count_100 = (df[["tasa_positividad"]].value_counts().get(100, 0))
print("Juegos con 100% de tasa de positividad:", count_100)

df[["tasa_positividad"]].hist(bins=30)
plt.title("Distribución tasa positividad")
plt.xlabel("Tasa positividad")
plt.show()

#TODO quitar el -1 de la grafica.
```

## C. Análisis de variables

**Objetivo:** Analizar los géneros más frecuentes en oferta frente a su volumen estimado de jugadores. Analizar estadísticamente la distribución de precios en la plataforma.

```python
# Desempaquetado de géneros y cálculo de frecuencias absolutas y relativas.
# Ej. Generar gráfico de barras del Top 10 de géneros.
# Escribe tu código aquí:

# Explode expande cada elemento de las listas, haciendo una fila por cada elemento y manteniendo el id al que pertenecia.
abs = df["Genres"].explode().value_counts()

# Frecuencias relativas. Normalice a true da jussto eso. Y *100 para que sea porcentaje.
rel = df["Genres"].explode().value_counts(normalize=True) * 100

print("[M]----------Top 10 Frecuencias Relativas----------")
print(rel.head(10).round(2))

top_10_absoluta = abs.head(10)

# Grafico de barras.
plt.bar(top_10_absoluta.index,top_10_absoluta.values)
plt.title("Top 10 generos")
plt.xlabel("Generos")
plt.ylabel("N juegos")
plt.show()
```

## D. Análisis de correlaciones y serie temporal

**Objetivo:** Estudiar la relación cruzada entre el precio y el éxito del videojuego. Analizar la serie temporal de lanzamientos de juegos por año en Steam.

```python
# Calcular la matriz de correlación (ej. Spearman) entre precio, positividad y número de reseñas.
# Ej. Generar visualización (Heatmap o Scatter plot con tendencia).
# Escribe tu código aquí:

# Modificamos la columna de Estimated Owners por el valor medio del intervalo dado.
def rangeToValue(val):
    
    # El formato es "X - Y".

    #Si no es valor o string devolvemos NaN.
    if pd.isna(val) or not isinstance(val, str):
        return np.nan
    
    # Comprobamos si cumple el formato.
    if '-' in val:
        values = val.split('-') # Separamos por el guion.
        low = float(values[0].strip()) # Quitamos el valor por debajo.
        high = float(values[1].strip()) # Quitamos el valor por encima.
        return (low + high) / 2.0 # Devolvemos el valor medio.

    return np.nan


# Nueva columna con los valores medios de los propietarios.
df["ownersMean"] = df["Estimated owners"].apply(rangeToValue)

# Guardamos los juegos que solo tengna la tasa de positividad valida y sus columnas para la correlacion que queramos.
aux = df[df["tasa_positividad"] != notEnough][["Price", "tasa_positividad", "totalReviews", "ownersMean"]]
if printPruebas:
    print(aux)

# Correlacion de Spearman.
correlation = aux.corr(method='spearman')
if printPruebas:
    print(correlation.round(2))

# Mapa de calor:
sns.heatmap(
    correlation, # Valores.
    annot=True, # Muestra los valores numeritos en cada cuadrado.
    cmap="coolwarm", # Colores de seaborn para que se vea mejor que con los por defecto.
    vmin=-1, vmax=1, # Rangos.
)

plt.title("Matriz correlacion entre precio, positividad y numero de reseñas.")
plt.show()
```

```python
# TODO: Extraer el año de lanzamiento de 'Release date' y analizar la evolución del catálogo de Steam año a año.
# Ej. Generar gráfico de línea/área.
# Escribe tu código aquí:

# Cogemos las fehcas del dataframe.
dates = df["Release date"]
if printPruebas:
    print("[P]----------Inicio Pruebas----------")
    print("[P]----------Comprobar que las fechas tienen formato fecha----------")
    print(dates)

# Cogemos solo los anyos de las fechas y la cantidad de cada anyo que hay.
years = df["Release date"].dt.year.value_counts()
if printPruebas:
    print("[P]----------Comprobar que solo ha cogido los anyos----------")
    print(years.head(5))
    print("[P]----------Comprobar los valores----------")
    print(years.values)
    print("[P]----------Comprobar los indices----------")
    print(years.index)

# Ordenamos los elementos de la serie.
years = years.sort_index(ascending=False)
if printPruebas:
    print("[P]----------Comrpobar que se han ordenado----------")
    print(years)

# Filtramos datos. Quitamos los anyios anteriores a la creacion de Steam y los anyios posteiores al 2026.
# Se excluye el 2026 porque al no haber acabado, los datos no estan completos.
years = years[(years.index < 2026) & (years.index >= 2003)]
if printPruebas:
    print("[P]----------Comprobar que se ha quitado el 2026----------")
    print(years)
    print("[P]----------Fin Pruebas----------")

# Grafica de lineas.
plt.plot(years)
plt.xticks(ticks = years.index, rotation = 45) # Para que se puedan ver todos los anyos mejor.
plt.title("Juegos lanzados por año")
plt.xlabel("Años")
plt.ylabel("Cantidad")
plt.show()
```
